```python
import json
with open('/home/jhpark/page_reproducibility/03_prior_perturbseq_compare/intermediate/priorprior_gsea_cluster.json', 'r') as f:
    prior_clusters = json.load(f)
with open('/home/jhpark/page_reproducibility/03_prior_perturbseq_compare/intermediate/priorpert_cofunc_gsea_cluster.json', 'r') as f:
    cofunc_clusters = json.load(f)
with open('/home/jhpark/page_reproducibility/03_prior_perturbseq_compare/intermediate/priorpert_coreg_gsea_cluster.json', 'r') as f:
    coreg_clusters = json.load(f)
```

```python
import numpy as np
plot_data = np.zeros((12, 6))
for i in range(1,13):
    for j in range(1, 7):
        terms1 = prior_clusters[f'C{i}']
        terms2 = cofunc_clusters[f'C{j}']
        jaccard = len(set(terms1).intersection(set(terms2)))/len(set(terms1).union(set(terms2)))
        plot_data[i-1,j-1] = jaccard
    
import pandas as pd
plot_df = pd.DataFrame(plot_data, index=[f'C{i}' for i in range(1, 13)], columns=[f'C{i}' for i in range(1, 7)]).T

import seaborn as sns
sns.heatmap(plot_df, vmin=0, vmax=1, cmap='Blues', annot=True, square=True)

import matplotlib.pyplot as plt
plt.ylabel('Prior-Perturbseq\n(Co-function) Clusters')
plt.xlabel('Prior-Prior Clusters')
plt.savefig('../figure_out/Fig.3F_1_terms_vs_terms.pdf')
```
